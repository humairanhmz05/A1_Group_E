The datasets contain historical records from 1970 to 2024. Missing values are mainly concentrated in older years, particularly in the poverty dataset. Older records also have incomplete state coverage because of historical geographical changes and sampling limitations. For example, the income dataset covers Peninsular Malaysia only in 1970 and 1974, while data for Labuan and Putrajaya are available only from 2007. The selected survey years contain all 16 states in both datasets, providing consistent state coverage for analysis. However, income values are nominal and have not been adjusted for inflation. Therefore, changes in income over time should be interpreted with this limitation in mind.

### Task 3: Income and Poverty Analysis

#### 3.1 Focus State Analysis — Terengganu

```python
# Get median income for Terengganu
terengganu_index = STATES.index("Terengganu")

terengganu_median = MEDIAN[terengganu_index, :]

print("Years:", SURVEY_YEARS)
print("Terengganu median income:", terengganu_median)
```

```python

# Calculate consecutive changes in median income
income_change = np.diff(terengganu_median)

# Calculate percentage changes
percentage_change = (
    income_change / terengganu_median[:-1]
) * 100

# Calculate change per year
year_gaps = np.diff(SURVEY_YEARS)
change_per_year = income_change / year_gaps

print("Income change:", income_change)
print("Percentage change (%):", percentage_change)
print("Change per year:", change_per_year)
```

```python

# Identify periods when median income decreased
income_fell = income_change < 0

print("Income fell:", income_fell)

# Display the periods when income fell
for i in range(len(income_fell)):
    if income_fell[i]:
        print(
            f"Income fell from {SURVEY_YEARS[i]} "
            f"to {SURVEY_YEARS[i + 1]}"
        )
```

```python

# Get mean income for Terengganu
terengganu_mean = MEAN[terengganu_index, :]

# Calculate mean-median gap
mean_median_gap = terengganu_mean - terengganu_median

print("Terengganu mean income:", terengganu_mean)
print("Mean-median gap:", mean_median_gap)
```

The median household income in Terengganu increased from RM4,694 in 2016 to RM6,627 in 2024. However, it decreased by RM755 (13.62%) from 2019 to 2020 before increasing again in the following periods. The mean income was higher than the median income in all selected years, with the largest mean–median gap of RM1,370 in 2022.

```python


# Focus states for all group members
FOCUS_STATES = [
    "Terengganu",
    "Pahang",
    "Sarawak",
    "Selangor",
    "W.P. Labuan"
]

year_gaps = np.diff(SURVEY_YEARS)

for state in FOCUS_STATES:
    state_index = STATES.index(state)

    median = MEDIAN[state_index, :]
    mean = MEAN[state_index, :]

    # Calculate income changes
    change = np.diff(median)
    percentage_change = change / median[:-1] * 100
    change_per_year = change / year_gaps

    # Boolean mask for falling income
    income_fell = change < 0

    # Calculate mean-median gap
    gap = mean - median

    print(f"\n{state}")

    # Label every result by state and period
    for i in range(len(change)):
        period = f"{SURVEY_YEARS[i]}-{SURVEY_YEARS[i + 1]}"

        print(
            f"{state} {period}: "
            f"change = RM{change[i]:.2f}, "
            f"percentage change = {percentage_change[i]:.2f}%, "
            f"change per year = RM{change_per_year[i]:.2f}, "
            f"income fell = {income_fell[i]}"
        )

    print("Mean - median gap by year:")
    for i, year in enumerate(SURVEY_YEARS):
        print(f"{state} {year}: RM{gap[i]:.2f}")

```

```python

# Index all original poverty records
poverty_index = index_by_key(poverty_records)

# Get absolute poverty values for 2016 and 2019
poverty_2016 = np.array([
    to_float(poverty_index[(state, 2016)]["poverty_absolute"])
    for state in STATES
])

poverty_2019 = np.array([
    to_float(poverty_index[(state, 2019)]["poverty_absolute"])
    for state in STATES
])

# Calculate change for all 16 states at once
poverty_change = poverty_2019 - poverty_2016

# Count states where absolute poverty rose
poverty_rise = poverty_change > 0
rise_count = np.count_nonzero(poverty_rise)

for i, state in enumerate(STATES):
    print(
        f"{state} 2016-2019: "
        f"{poverty_change[i]:.2f} percentage points"
    )

print("Number of states with a rise:", rise_count)
```

All 16 states recorded an increase in absolute poverty between 2016 and 2019. However, this does not necessarily mean that poverty truly increased because Malaysia revised its Poverty Line Income (PLI) methodology in 2019. The revised methodology changed how minimum household needs were measured, including food and non-food requirements. Therefore, the sharp difference may partly reflect changes in the measurement method rather than a real increase in poverty. POVERTY_YEARS starts in 2019 to focus on data measured under the revised methodology.
