```python
STATES = sorted(set(record["state"] for record in income_records))
print(STATES)
print(len(STATES))
```

```python
matric_numbers=[314171,313819,313374,313994,313989]
names= ["Naurah", "Irdina", "Humaira", "Ainul", "Khairina"]
used_states=set()
GROUP_STATES=[None]*len(matric_numbers)

for i in sorted(range(len(matric_numbers)) ,
                key=lambda i: matric_numbers[i]):

  start = matric_numbers[i] % 16

  for offset in range(16): 
    state = STATES[(start + offset) % 16]
    if state not in used_states: 
        GROUP_STATES[i] = state 
        used_states.add(state)
        break
    
for name, matric, state in zip(names, matric_numbers, GROUP_STATES):
   print(name, matric, "->", state)
```

```python
print("INCOME DATASET")

income_fields = sorted({
    key
    for record in income_records
    for key in record
})

for field in income_fields:
    none_count = sum(
        record.get(field) is None
        for record in income_records
    )
    
    print(field, ":", none_count)
```

```python
print("POVERTY DATASET")

poverty_fields = sorted({
    key
    for record in poverty_records
    for key in record
})

for field in poverty_fields:
    none_count = sum(
        record.get(field) is None
        for record in poverty_records
    )
    
    print(field, ":", none_count)
```

```python
print("INCOME DATA TYPES")

for field in income_fields:
    types = set(
        type(record[field]).__name__
        for record in income_records
        if record[field] is not None
    )
    
    print(field, ":", types)
```

```python
print("POVERTY DATA TYPES")

for field in poverty_fields:
    types = set(
        type(record[field]).__name__
        for record in poverty_records
        if record[field] is not None
    )
    
    print(field, ":", types)
```

```python
income_survey_records = [
    record
    for record in income_records
    if get_year(record) in SURVEY_YEARS
]

print("Number of filtered income records:", len(income_survey_records))
```

```python
print("CHECKING INCOME STATES")

for year in SURVEY_YEARS:
    states_this_year = {
        record["state"]
        for record in income_survey_records
        if get_year(record) == year
    }

    missing_states = set(STATES) - states_this_year

    print(year, ":", len(states_this_year), "states")

    if missing_states:
        print("Missing states:", sorted(missing_states))
    else:
        print("All 16 states are present.")
```

```python
poverty_survey_records = [
    record
    for record in poverty_records
    if get_year(record) in POVERTY_YEARS
]

print(
    "Number of filtered poverty records:",
    len(poverty_survey_records)
)

print()

print("CHECKING POVERTY STATES")

for year in POVERTY_YEARS:
    states_this_year = {
        record["state"]
        for record in poverty_survey_records
        if get_year(record) == year
    }

    missing_states = set(STATES) - states_this_year

    print(year, ":", len(states_this_year), "states")

    if missing_states:
        print("Missing states:", sorted(missing_states))
    else:
        print("All 16 states are present.")
```

```python
print("INCOME DATASET: NONE VALUES BY YEAR")

for year in sorted(set(get_year(record) for record in income_records)):
    records_this_year = [
        record for record in income_records
        if get_year(record) == year
    ]

    none_count = sum(
        record.get(field) is None
        for record in records_this_year
        for field in record
    )

    if none_count > 0:
        print(year, ":", none_count, "None values")


print()
print("POVERTY DATASET: NONE VALUES BY YEAR")

for year in sorted(set(get_year(record) for record in poverty_records)):
    records_this_year = [
        record for record in poverty_records
        if get_year(record) == year
    ]

    none_count = sum(
        record.get(field) is None
        for record in records_this_year
        for field in record
    )

    if none_count > 0:
        print(year, ":", none_count, "None values")
```

```python
print("INCOME DATASET: MISSING STATES BY YEAR")

for year in sorted(set(get_year(record) for record in income_records)):
    states_this_year = {
        record["state"]
        for record in income_records
        if get_year(record) == year
    }

    missing_states = set(STATES) - states_this_year

    if missing_states:
        print(year, ":", sorted(missing_states))


print()
print("POVERTY DATASET: MISSING STATES BY YEAR")

for year in sorted(set(get_year(record) for record in poverty_records)):
    states_this_year = {
        record["state"]
        for record in poverty_records
        if get_year(record) == year
    }

    missing_states = set(STATES) - states_this_year

    if missing_states:
        print(year, ":", sorted(missing_states))
```

The datasets contain historical records from 1970 to 2024. Missing values are mainly concentrated in older years, particularly in the poverty dataset. Older records also have incomplete state coverage because of historical geographical changes and sampling limitations. For example, the income dataset covers Peninsular Malaysia only in 1970 and 1974, while data for Labuan and Putrajaya are available only from 2007. The selected survey years contain all 16 states in both datasets, providing consistent state coverage for analysis. However, income values are nominal and have not been adjusted for inflation. Therefore, changes in income over time should be interpreted with this limitation in mind.
